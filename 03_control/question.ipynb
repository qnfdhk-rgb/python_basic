문제 5: 집 구하기

문제 6: 직장 찾기
"카피바라는 직장을 구해야 한다!"
카피바라는 서울에서 일자리를 찾고 있습니다. 직장에 따라 일의 강도가 다릅니다.
사용자에게 "원하는 직장 종류를 입력하세요 (프리랜서/중소기업/대기업): "를 입력받으세요.
입력값이 '프리랜서'이면 "시간은 자유롭지만 불안정하다."를 출력하세요.
입력값이 '중소기업'이면 "안정적이지만 일이 많다."를 출력하세요.
그 밖에 입력값에 대해서는 “좋은 직장을 가지고 계시는군요”를 출력하세요.

```python
work = input('원하는 직장 종류를 입력하세요(프리랜서/중소기업/대기업): ')

if work == '프리랜서':
    print('시간은 자유롭지만 불안정하다')
elif work == '중소기업':
    print('안정적이지만 일이 많다.')
else:
    print('좋은 직장을 가지고 계시는군요.')
```

문제 7: 생활 루틴 정하기
"카피바라는 서울에서 생활 루틴을 정해야 한다."
서울 생활에 적응한 카피바라는 하루 루틴을 정하려 합니다.
사용자에게 "아침에 일어나는 시간을 입력하세요 (숫자 입력, 6~10): "를 입력받으세요.
입력값이 6이면 "운동을 한다."를 출력하세요.
입력값이 7이면 "공부를 한다."를 출력하세요.
그 외에는 "바로 출근 준비를 한다."를 출력하세요.

```python
routine = int(input("아침에 일어나는 시간을 입력하세요(숫자 입력, 6~10): "))

if routine == 6:
    print("운동을 한다.")
elif routine == 7:
    print("공부를 한다.")
else:
    print("바로 출근 준비를 한다.")
```

```python
time_pay = 1_000
work_hours = int(input("일한 시간을 입력하세요 (정수): "))

pay = 0
current_hour = 0

while current_hour < work_hours:
    pay += time_pay
    current_hour += 1
    print(f"오늘의 알바비는 {pay}원 입니다.")
```
