```python
abclist = ['A','B','C','D','E','F']
print(abclist)    # ['A', 'B', 'C', 'D', 'E', 'F']

del abclist[0]    # 0번째 인덱스의 요소 제거
print(abclist)    # ['B', 'C', 'D', 'E', 'F']

del abclist[1:3]  # 1번째 인덱스에서 3번째까지 제거
print(abclist)    # ['B', 'E', 'F']

del abclist[:]    # 모든 요소 제거
print(abclist)  
```

```python
from collections import deque

dq = deque()

dq.append("a")
dq.append("b")
dq.append("c")

print(dq)    # deque(['a', 'b', 'c'])
```

```python
dq.appendleft('x')
print(dq)   # deque(['x', 'a', 'b', 'c'])
```

```python
value = dq.pop()
print(value)    # c

left_value = dq.popleft()
print(left_value)    # x

print(dq)    # deque(['a', 'b'])
```
