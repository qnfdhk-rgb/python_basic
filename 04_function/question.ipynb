```python
def calculate_salary(base_salary, bonus):
    total_salary = base_salary + bonus # 총월급
    net_salary = total_salary * (1-0.1) # 실수령액
    tax = total_salary * 0.1 # 세금공제액
    return f'세전 월급: [{total_salary}]원, 세금: [{tax}]원, 실수령액: [{net_salary}]원'

calculate_salary(100_000_000, 50_000_000)
```

```python
g_variable = "global variable"

def modify_global():
    global g_variable
    g_variable = "global modified"

print(g_variable)    # global variable
modify_global()
print(g_variable)    # global modified
```

```python
def outer_function():
    variable = "outer variable"

    def inner_function():
        nonlocal variable
        variable = "modified"

    print(variable)
    inner_function()
    print(variable)

outer_function()
```

```python
class Capybara:
    # 클래스 속성 (모든 인스턴스가 공유함)
    species = "카피바라"

    # 생성자 (인스턴스 속성 설정)
    def __init__(self, name, current_job):
        self.name = name                # 인스턴스 속성 1
        self.current_job = current_job  # 인스턴스 속성 2

    # introduce 메서드 정의 (자기소개 문자열 반환)
    def introduce(self):
        return f"안녕하세요, 저는 {self.species}족 {self.name}입니다. 현재 {self.current_job}(으)로 일하고 있습니다."


# 2. 첫 번째 카피바라 인스턴스 생성 및 확인
capy1 = Capybara("사막여우", "개발자")
print(capy1.introduce())


# 3. 두 번째 카피바라 인스턴스 생성 및 독립성 확인
capy2 = Capybara("듬직이", "디자이너")
print(capy2.introduce())
```

```python
class Person:
    national = 'korea'

    def greeting(self):
        return "Hello. This is Python"

class Learner:
    def __init__(self, subject):
        self.subject = subject

    def learn(self):
        return 'I am learning ' + self.subject

class Student(Learner):
    def __init__(self, name, subject):
        super().__init__(subject)
        self.name = name

    def greeting(self):
        return 'Hello. my name is ' + self.name

student = Student('카피바라', 'python')
print(student.greeting())
print(student.learn())
```

```python
class Animal:
    def do(self):
        print('👉👈')
class Dog(Animal):
    def do(self):
        print('🐶🐶🐶')
class Duck(Animal):
    def do(self):
        print('🦆🦆🦆')
class Fish(Animal):
    pass    # do 호출시 부모 클래스 Animal#do가 호출된다.

for each in (Dog(), Duck(), Fish()):
    each.do()
```

```python
class Person:
    national = 'Korea'

    def introduce(self):
        return f'안녕하세요, 저는 한국에서 왔습니다.'
# 2. 자식 클래스 
class Candidate(Person):
    # 생성장에서 이름과 관심 분야를 매개변수로 받아 저장하는 부분
    def __init__(self, name, field):
        self.name = name
        self.field = field

    # 메서드 오버라이드 부분 (self 꼭 적어야함!!!)
    def introduce(self):
        return f'안녕하세요, 저는 {self.name}입니다. {self.field} 분야에 관심이 있습니다.'

# person1은 부모 class로 init가 없음 ()로 비워서 적용
# person1은 class Person, 2는 class Candidate
person1 = Person()
person2 = Candidate('pick', '디자이너')

# person2의 introduce()를 출력
print(person2.introduce())
```

```python
# class Mentor 이라는 설계도
class Mentor:
    # guide라는 메서드로 행동을 저장
    def guide(self):
        return f'이직 준비를 도와드립니다.'

# class Lecturer 이라는 설계도2
class Lecturer:
    # teach 이라는 메서드 행동 저장
    def teach(self):
        return f'이직에 필요한 기술을 가르쳐드립니다.'

# class Coach 이라는 설계도3
class Coach(Mentor, Lecturer):
    def support(self):
        return f'여러 방면으로 지원을 아끼지 않겠습니다.'

# person이라는 사람들의 직업 부여
person1 = Mentor()
person2 = Lecturer()
person3 = Coach()

# 각 사람의 메서드 출력
print(person1.guide())
print(person2.teach())
print(person3.support())
```
